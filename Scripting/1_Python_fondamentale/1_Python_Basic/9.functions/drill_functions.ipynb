### 4. Abbreviate a two-word name

Example : 

````python
def abbreviate_name(name):
    # Your code
    mots= name.split()
    crochet=""
    for n in mots:
        crochet += n[0].upper()
    return "sam harris"
````

### 5. Sum of positive 
You get an array of numbers, return the sum of all of the positives ones.

Example :  
` [1,-4,7,12] => 1 + 7 + 12 = 20`

````python
def positive_sum(numbers):
    total=0
    for n in numbers:
        if n > 0:
            total+= n
    return total
positive_sum([1, 7, 12, -4])
````

### 6. Sum mixed array

Example : 
```
mixed_sum(['5', '0', 9, 3, 2, 1, '9', 6, 7]) => 42
```

````python
def mixed_sum(array):
    total=0
    for p in array:
        total += int(p)
    return total
array=(["5", 5, "5", 5])
mixed_sum(array)
````
