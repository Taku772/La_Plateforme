### 8. Create a dictionary to build the price base of the products corresponding to the following table.

    Laser sword // 229.0
    Mitendo DX // 127.30
    Linux cushion // 74.50
    Goldorak briefs // 29.90
    Nextpresso station // 184.60

```python
dico={}
dico["Laser sword"] = 229.0 
dico["Mitendo DX"] = 127.30 
dico["Linux cushion"] = 74.50 
dico["Goldorak briefs"] = 29.90
dico["Nextpresso station"] = 184.60
print(dico)
```

### 9. Calculate the total price of the items of the dictionary.

```python
sommevalues= sum(dico.values())
print(sommevalues)
```

### 10. Remove one of the articles from the dictionary.

```python
if "Laser sword" in dico:
    del dico["Laser sword"]
print(dico)
```
